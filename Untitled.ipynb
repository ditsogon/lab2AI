```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
df = pd.read_csv("processed_intermediate_coreml.csv")

print("Размер датасета:", df.shape)
print("\nПервые пять строк:")
display(df.head())

print("\nНазвания столбцов:")
print(df.columns.tolist())

print("\nПропущенные значения:")
print(df.isnull().sum().sum())
```

```python
# Признаки, по которым модель будет делать прогноз
X = df.drop(columns=["customer_id", "target_reg", "churn"])

# Целевая переменная для регрессии
y_reg = df["target_reg"]

# Разделяем данные: 75% для обучения, 25% для проверки
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X,
    y_reg,
    test_size=0.25,
    random_state=42
)

print("Размер обучающей выборки:", X_train_reg.shape)
print("Размер тестовой выборки:", X_test_reg.shape)
```

```python
# Создаём модель линейной регрессии
reg_model = LinearRegression()

# Обучаем модель на обучающей выборке
reg_model.fit(X_train_reg, y_train_reg)

# Делаем прогноз на тестовой выборке
y_pred_reg = reg_model.predict(X_test_reg)

print("Первые 10 реальных значений:")
print(y_test_reg.iloc[:10].values)

print("\nПервые 10 предсказанных значений:")
print(np.round(y_pred_reg[:10], 2))
```

```python
mse = mean_squared_error(y_test_reg, y_pred_reg)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

print("Оценка модели регрессии:")
print("MSE:", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("MAE:", round(mae, 2))
print("R²:", round(r2, 4))
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(y_test_reg, y_pred_reg, alpha=0.3)
plt.xlabel("Реальные значения")
plt.ylabel("Предсказанные значения")
plt.title("Реальные и предсказанные значения регрессии")

# Линия идеального прогноза
min_value = min(y_test_reg.min(), y_pred_reg.min())
max_value = max(y_test_reg.max(), y_pred_reg.max())
plt.plot([min_value, max_value], [min_value, max_value], "r--")

plt.show()
```

```python
# Целевая переменная классификации
y_clf = df["churn"]

# Разделяем данные на обучающую и тестовую выборки
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X,
    y_clf,
    test_size=0.25,
    random_state=42,
    stratify=y_clf
)

print("Обучающая выборка:", X_train_clf.shape)
print("Тестовая выборка:", X_test_clf.shape)

print("\nРаспределение классов в исходных данных:")
print(y_clf.value_counts())

print("\nДоли классов:")
print(y_clf.value_counts(normalize=True).round(3))
```

```python
# Создаём модель с масштабированием признаков
# и компенсацией дисбаланса классов
clf_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(
        class_weight="balanced",
        max_iter=2000,
        random_state=42
    )
)

# Обучаем модель
clf_model.fit(X_train_clf, y_train_clf)

# Получаем прогнозы
y_pred_clf = clf_model.predict(X_test_clf)

print("Первые 20 реальных классов:")
print(y_test_clf.iloc[:20].values)

print("\nПервые 20 предсказанных классов:")
print(y_pred_clf[:20])
```

```python
accuracy = accuracy_score(y_test_clf, y_pred_clf)
precision = precision_score(y_test_clf, y_pred_clf, zero_division=0)
recall = recall_score(y_test_clf, y_pred_clf, zero_division=0)
f1 = f1_score(y_test_clf, y_pred_clf, zero_division=0)

print("Оценка модели классификации:")
print("Accuracy:", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1-score:", round(f1, 4))

print("\nМатрица ошибок:")
print(confusion_matrix(y_test_clf, y_pred_clf))

print("\nОтчёт классификации:")
print(classification_report(y_test_clf, y_pred_clf, zero_division=0))
```
